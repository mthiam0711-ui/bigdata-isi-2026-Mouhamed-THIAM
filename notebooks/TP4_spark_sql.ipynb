## Partie A — Mise en place (15 min)

### A.1 — Charger les quatre sources et créer les vues

Chargez `customers.csv`, `orders.csv`, `products.csv` (CSV : `header=True`, `inferSchema=True`) et `payments.json`, puis créez les vues temporaires `clients`, `commandes`, `produits`, `paiements`.

## Partie C — L'enquête FCFA (30 min)

### C1 — Le symptôme : la somme naïve
Calculez le CA total directement sur la colonne brute, et **notez le résultat**.

````python
# === À COMPLÉTER ===
ca_naif = spark.sql("""
    SELECT SUM(montant_total_fcfa) AS ca FROM commandes
""").first()["ca"]
releves["C1_ca_naif"] = ca_naif
print(f"CA naif : {ca_naif:,.0f}")
````

### C2 — Diagnostiquer
Comptez les valeurs de `montant_total_fcfa` qui ne sont **pas** de purs nombres, puis affichez 10 valeurs fautives distinctes.

````python
# === À COMPLÉTER ===
nb_pollues = spark.sql("""
    SELECT COUNT(*) AS nb
    FROM commandes
    WHERE montant_total_fcfa NOT rlike '^[0-9]+$'
""").first()["nb"]
releves["C2_nb_valeurs_polluees"] = nb_pollues
print("valeurs polluees :", nb_pollues)

spark.sql("""
    SELECT DISTINCT montant_total_fcfa
    FROM commandes
    WHERE montant_total_fcfa NOT rlike '^[0-9]+$'
    LIMIT 10
""").show(truncate=False)
````

**Question C** — Expliquez en deux phrases pourquoi la requête C1 rend un résultat **faux sans lever d'erreur**.

*Votre réponse :*

`SUM(montant_total_fcfa)` porte sur une colonne `string` : Spark tente une conversion implicite en nombre pour pouvoir sommer, mais les valeurs polluées comme `"10000 FCFA"` ne correspondent à aucun nombre valide et sont donc silencieusement transformées en `NULL` (pas d'exception levée, mode non-ANSI par défaut). Comme `SUM` ignore les `NULL`, le résultat est calculé quand même, mais il sous-estime le vrai chiffre d'affaires puisqu'une partie des commandes est purement et simplement exclue du total sans aucun signal d'erreur.

### C3 — Nettoyer : la vue `commandes_clean`
Complétez la regex : supprimer **tout ce qui n'est pas un chiffre**, puis caster en `BIGINT`. On conserve la colonne brute sous `montant_raw`.

````python
# === À COMPLÉTER ===
spark.sql("""
    CREATE OR REPLACE TEMP VIEW commandes_clean AS
    SELECT order_id, customer_id, date_commande, statut, canal,
           frais_livraison_fcfa,
           montant_total_fcfa AS montant_raw,
           CAST(regexp_replace(trim(montant_total_fcfa),
                '[^0-9]', '') AS BIGINT) AS montant_fcfa
    FROM commandes
""")
spark.sql("SELECT montant_raw, montant_fcfa FROM commandes_clean LIMIT 5").show()
````

### C4 — Valider : mesurer, pas affirmer
Vérifiez qu'aucun `NULL` n'a été produit, contrôlez `MIN`/`MAX`, et recalculez le CA.

````python
# === À COMPLÉTER ===
validation = spark.sql("""
    SELECT COUNT(*)                        AS nb_lignes,
           COUNT(montant_fcfa)             AS nb_castes,
           COUNT(*) - COUNT(montant_fcfa)  AS nb_null,
           MIN(montant_fcfa)               AS mini,
           MAX(montant_fcfa)               AS maxi,
           SUM(montant_fcfa)               AS ca_total
    FROM commandes_clean
""")
validation.show()
releves["C4_ca_nettoye"] = validation.first()["ca_total"]
````

### C5 — La preuve chiffrée
Calculez l'écart entre le CA naïf (C1) et le CA nettoyé (C4), en FCFA et en pourcentage.

````python
# === À COMPLÉTER ===
ecart = releves["C4_ca_nettoye"] - releves["C1_ca_naif"]
releves["C5_ecart_fcfa"] = ecart
releves["C5_ecart_pct"]  = 100 * ecart / releves["C1_ca_naif"]
print(f"Ecart : {ecart:,.0f} FCFA soit {releves['C5_ecart_pct']:.2f} %")
````

## Partie D — Les indicateurs de la direction (40 min)

Toutes les requêtes portent sur `commandes_clean` (et `paiements` pour D5). Après chaque résultat, ajoutez **une phrase d'interprétation métier** dans la cellule markdown qui suit.

### D1 — CA et commandes par statut
Quelle part du CA est réellement `livree` ?

````python
# === À COMPLÉTER ===
spark.sql("""
    SELECT statut,
           COUNT(*)          AS nb_commandes,
           SUM(montant_fcfa) AS ca_fcfa
    FROM commandes_clean
    GROUP BY statut
    ORDER BY ca_fcfa DESC
""").show()

releves["D1_part_ca_livree_pct"] = spark.sql("""
    SELECT 100.0 * SUM(CASE WHEN statut = 'livrée' THEN montant_fcfa ELSE 0 END)
                 / SUM(montant_fcfa) AS pct
    FROM commandes_clean
""").first()["pct"]
````

*Votre réponse :*

Une part significative du CA correspond à des commandes qui ne sont pas (encore, ou plus) livrées : `en_cours`, `annulée` et `retournée` pèsent aussi lourd dans les montants agrégés. Seul le CA des commandes au statut `livrée` correspond à un revenu réellement encaissé et non susceptible d'être annulé ; piloter l'activité sur le CA « brut » toutes commandes confondues surestimerait donc la performance réelle de l'entreprise.

### D2 — Le CA mensuel des commandes livrées
`date_trunc('month', ...)`, tri chronologique. Repérez la tendance et le mois record.

````python
# === À COMPLÉTER ===
ca_mensuel = spark.sql("""
    SELECT date_trunc('month', date_commande) AS mois,
           COUNT(*)                           AS nb_commandes,
           SUM(montant_fcfa)                  AS ca_fcfa
    FROM commandes_clean
    WHERE statut = 'livrée'
    GROUP BY date_trunc('month', date_commande)
    ORDER BY mois
""")
ca_mensuel.show(24, truncate=False)
releves["D2_mois_record"] = spark.sql("""
    SELECT date_trunc('month', date_commande) AS mois,
           SUM(montant_fcfa)                  AS ca_fcfa
    FROM commandes_clean
    WHERE statut = 'livrée'
    GROUP BY date_trunc('month', date_commande)
    ORDER BY ca_fcfa DESC
    LIMIT 1
""").first()["mois"]
````

*Votre réponse :*

Le CA mensuel des commandes livrées est globalement en croissance sur toute la période (environ 293M FCFA en juillet 2024 contre plus de 430M FCFA au printemps 2026), avec deux pics nets en décembre 2024 (490M) et surtout **décembre 2025 (634M FCFA, le mois record)** — cohérent avec un effet saisonnier de fin d'année (fêtes). En dehors de ces pics de décembre, la tendance de fond reste une croissance progressive du nombre de commandes et du CA d'un mois sur l'autre.

### D3 — Panier moyen par canal
`ROUND(AVG(montant_fcfa), 0)` — mobile ou web, qui dépense le plus par commande ?

````python
# === À COMPLÉTER ===
panier = spark.sql("""
    SELECT canal,
           COUNT(*)                    AS nb_commandes,
           SUM(montant_fcfa)           AS ca_fcfa,
           ROUND(AVG(montant_fcfa), 0) AS panier_moyen_fcfa
    FROM commandes_clean
    GROUP BY canal
    ORDER BY ca_fcfa DESC
""")
panier.show()
releves["D3_panier_moyen_mobile"] = panier.filter(panier.canal == "mobile_app").first()["panier_moyen_fcfa"]
````

*Votre réponse :*

Le panier moyen est en réalité presque identique entre les deux canaux (233 457 FCFA pour `mobile_app` contre 231 878 FCFA pour `web`, soit moins de 1% d'écart) : le canal mobile ne dépense donc pas davantage par commande. En revanche, `mobile_app` génère près du double du CA total du canal web, uniquement parce qu'il concentre beaucoup plus de commandes (32 519 contre 17 481) — c'est un canal dominant par le **volume**, pas par le panier.

### D4 — Top clients, avec HAVING
Top 10 des clients par CA (`GROUP BY customer_id`), en ne gardant que les clients dépassant **1 000 000 FCFA** de CA cumulé. Un client sort-il du lot ?

````python
# === À COMPLÉTER ===
spark.sql("""
    SELECT customer_id,
           COUNT(*)          AS nb_commandes,
           SUM(montant_fcfa) AS ca_fcfa
    FROM commandes_clean
    GROUP BY customer_id
    HAVING SUM(montant_fcfa) > 1000000
    ORDER BY ca_fcfa DESC
    LIMIT 10
""").show()
````

*Votre réponse :*

Oui, un client sort nettement du lot : **C000001** avec 3143 commandes et 732 953 000 FCFA de CA cumulé, contre seulement 421 commandes et ~95 M FCFA pour le second client du classement — soit près de 7,5 fois plus de commandes et 7,7 fois plus de CA. Ce profil ne ressemble pas à un client particulier « normal » mais plutôt à un compte grossiste/revendeur, ce qui justifierait de le traiter à part dans les analyses (sinon il fausse fortement toute moyenne par client).

### D5 — Paiements par méthode
Nombre et pourcentage par méthode. Quelle part totale pour le **mobile money** (Orange Money + Wave + Free Money) ?

````python
# === À COMPLÉTER ===
spark.sql("""
    SELECT methode,
           COUNT(*) AS nb,
           ROUND(100 * COUNT(*) / SUM(COUNT(*)) OVER (), 1) AS part_pct
    FROM paiements
    GROUP BY methode
    ORDER BY nb DESC
""").show()

releves["D5_part_mobile_money_pct"] = spark.sql("""
    SELECT ROUND(100.0 * SUM(CASE WHEN methode IN ('Orange Money', 'Wave', 'Free Money')
                                  THEN 1 ELSE 0 END) / COUNT(*), 1) AS pct
    FROM paiements
""").first()["pct"]
````

*Votre réponse :*

Le mobile money (Orange Money 35,1% + Wave 30,1% + Free Money 5,0%) représente **70,2%** des paiements, largement devant le paiement à la livraison (22,0%) et la carte bancaire (7,9%). Cela confirme la place centrale du mobile money dans les habitudes de paiement e-commerce au Sénégal, et suggère de prioriser la fiabilité/l'intégration de ces opérateurs (Orange Money, Wave) dans le parcours de paiement.

## Partie E — SQL ou API ? `explain()` tranche (20 min)

### E1 — D3 en API DataFrame
Réécrivez le panier moyen par canal avec `groupBy().agg()`. Les chiffres doivent être **identiques**.

````python
from pyspark.sql import functions as F

# === À COMPLÉTER ===
commandes_clean_df = spark.table("commandes_clean")

panier_api = (commandes_clean_df
    .groupBy("canal")
    .agg(
        F.count("*").alias("nb_commandes"),
        F.sum("montant_fcfa").alias("ca_fcfa"),
        F.round(F.avg("montant_fcfa"), 0).alias("panier_moyen_fcfa"),
    )
    .orderBy(F.col("ca_fcfa").desc()))
panier_api.show()
````

### E2 — B4 en API
La même requête « livrées / mobile / décembre 2025 », version `filter`.

````python
# === À COMPLÉTER ===
nb_api = (spark.table("commandes")
    .filter((F.col("statut") == "livrée") &
            (F.col("canal") == "mobile_app") &
            (F.year("date_commande") == 2025) &
            (F.month("date_commande") == 12))
    .count())
print(nb_api)
````

### E3 — Comparer les plans
Affichez le plan physique de la version SQL de D3 et de `panier_api`.

````python
panier_sql = spark.sql("""
    SELECT canal, COUNT(*) AS nb_commandes,
           SUM(montant_fcfa) AS ca_fcfa,
           ROUND(AVG(montant_fcfa), 0) AS panier_moyen_fcfa
    FROM commandes_clean
    GROUP BY canal
    ORDER BY ca_fcfa DESC
""")
panier_sql.explain()
panier_api.explain()
````

**Question E** — Les plans physiques sont-ils identiques ? Où voit-on le filtre poussé vers la lecture (`PushedFilters` / `Filter` près du `FileScan`) ? Que concluez-vous sur le choix SQL vs API ? (3 phrases)

*Votre réponse :*

Les deux plans physiques sont **strictement identiques** dans leur structure : même `FileScan csv` (ne lisant que les deux colonnes nécessaires `canal` et `montant_total_fcfa`, grâce à la column pruning), même `Project` appliquant `trim` + `regexp_replace` + `cast`, mêmes deux `HashAggregate` (partiel puis final) et le même `Sort` final. Comme la requête D3 n'a pas de clause `WHERE`, on ne voit ni `Filter` ni entrée dans `PushedFilters` (qui reste vide `[]`) dans aucune des deux versions — il n'y a simplement rien à pousser vers la lecture ici. On en conclut que l'API DataFrame et le SQL sont deux syntaxes qui passent par le **même optimiseur Catalyst** et produisent le même plan d'exécution : le choix entre les deux est donc une question de lisibilité/préférence d'équipe, pas de performance.

## Quiz éclair (10 min)

1. Que retourne `spark.sql(...)` : une liste, un DataFrame ou un fichier ?
2. `createOrReplaceTempView` copie-t-elle les données ? Quelle est la portée de la vue ?
3. Pourquoi `WHERE email = NULL` ne renvoie-t-il jamais rien ?
4. `SUM` sur une colonne `string` polluée : erreur ou résultat faux ? Pourquoi est-ce dangereux ?
5. `WHERE` et `HAVING` : lequel filtre les groupes, lequel filtre les lignes ?

*Notez votre score dans la cellule suivante.*

*Votre réponse :*

1. `spark.sql(...)` retourne un **DataFrame** (paresseux / lazy) — ni une liste ni un fichier ; il faut une action (`.show()`, `.collect()`, `.first()`...) pour déclencher réellement le calcul.
2. Non, `createOrReplaceTempView` ne copie pas les données : elle enregistre juste un alias logique (un plan) vers le DataFrame source dans le catalogue Spark. Sa portée est limitée à la **SparkSession courante** (elle disparaît si la session/le notebook redémarre, et n'est pas visible d'une autre session).
3. `WHERE email = NULL` ne renvoie jamais rien car en SQL, `NULL` représente une valeur inconnue : toute comparaison (`=`, `<`, `>`...) avec `NULL` est elle-même `NULL` (ni vrai ni faux), donc jamais retenue par un `WHERE`. Il faut utiliser `IS NULL`.
4. `SUM` sur une colonne `string` polluée ne lève pas d'erreur (mode non-ANSI) : les valeurs non convertibles deviennent `NULL` et sont ignorées, ce qui donne un **résultat faux mais silencieux** — c'est dangereux car rien n'indique visuellement que le calcul est erroné, contrairement à une exception qui aurait alerté immédiatement.
5. `WHERE` filtre les **lignes** avant l'agrégation ; `HAVING` filtre les **groupes** (résultats de `GROUP BY`) après l'agrégation.
